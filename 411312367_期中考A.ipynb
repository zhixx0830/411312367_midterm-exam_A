````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
def init_db():
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()

        # 每次執行都重建資料表
        cursor.execute("DROP TABLE IF EXISTS activities")
        cursor.execute("DROP TABLE IF EXISTS registrations")

        # 建立活動表
        cursor.execute('''
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT UNIQUE,
                date TEXT,
                quota INTEGER
            )
        ''')

        # 建立報名紀錄表
        cursor.execute('''
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT,
                activity_name TEXT,
                pax INTEGER,
                status TEXT
            )
        ''')

        # 寫入初始資料
        initial_activities = [
            ('迎新晚會', '2026-11-20', 6),
            ('程式工作坊', '2026-11-25', 4),
            ('籃球友誼賽', '2026-11-28', 3)
        ]

        cursor.executemany("INSERT INTO activities (name, date, quota) VALUES (?, ?, ?)", initial_activities)
        conn.commit()
    finally:
        conn.close()

init_db()
print("✅ 資料庫與初始資料已建立")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    conn = sqlite3.connect('activity_system.db')
    try:
        print("【活動表 activities】")
        df_activities = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n【報名紀錄表 registrations】")
        df_registrations = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_registrations)
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
ai_cache = {}

def parse_intent_with_ai(user_input):
    if user_input in ai_cache:
        return ai_cache[user_input]

    prompt = f"""
    你是一個系統意圖解析器。請從使用者的話中擷取「動作」、「活動」與「人數」。
    - 動作限制：報名、取消、查詢、統計、未知。
    - 活動限制：迎新晚會、程式工作坊、籃球友誼賽、未知。
    - 人數限制：請提取阿拉伯數字（包含朋友、攜伴、我自己等，如「我和兩個朋友」為3），若無明確數字則預設為1。
    - 若使用者問「各活動狀況」、「報名狀況」，動作請填「統計」，活動填「未知」。

    使用者說：{user_input}
    請嚴格只輸出逗號分隔的格式，例如：報名,迎新晚會,2
    """

    retries = 3
    for attempt in range(retries):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_reply = response.choices[0].message.content.strip()
            time.sleep(2) # 避免呼叫過快

            # 解析回應並驗證合法性
            parts = raw_reply.split(',')
            action = parts[0].strip() if len(parts) > 0 else "未知"
            activity = parts[1].strip() if len(parts) > 1 else "未知"

            # 確保是在允許清單內
            allowed_actions = ["報名", "取消", "查詢", "統計"]
            allowed_activities = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

            if action not in allowed_actions: action = "未知"
            if activity not in allowed_activities: activity = "未知"

            # 解析人數 (Regex 找出數字)
            pax = 1
            if len(parts) > 2:
                nums = re.findall(r'\d+', parts[2])
                if nums:
                    pax = int(nums[0])

            result = (action, activity, pax, raw_reply)
            ai_cache[user_input] = result
            return result

        except Exception as e:
            if "429" in str(e):
                print(f"⚠️ 遇到 API 限制，等待 30 秒後重試... (第 {attempt+1}/{retries} 次)")
                time.sleep(30)
            else:
                print(f"AI 呼叫發生錯誤: {e}")
                return ("未知", "未知", 1, "錯誤")

    return ("未知", "未知", 1, "超過重試次數")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
test_sentences = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),
    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),
    ("籃球友誼賽還有名額嗎", "查詢", "籃球友誼賽", 1),
    ("我和兩個朋友要報名籃球賽", "報名", "籃球友誼賽", 3),
    ("幫我跟我女朋友報名迎新", "報名", "迎新晚會", 2),
    ("各活動報名狀況", "統計", "未知", 1),
    ("我要吃漢堡", "未知", "未知", 1),
    ("幫我報名唱歌比賽", "報名", "未知", 1)
]

passed = 0
for text, exp_action, exp_act, exp_pax in test_sentences:
    act, obj, pax, raw = parse_intent_with_ai(text)
    match = (act == exp_action) and (obj == exp_act) and (pax == exp_pax)
    if match: passed += 1

    print(f"輸入: {text}")
    print(f"  AI 原始回覆: {raw}")
    print(f"  系統解讀: 動作={act}, 活動={obj}, 人數={pax}")
    print(f"  結果: {'✅ 通過' if match else '❌ 失敗'}\n")

print(f"總結：通過 {passed} / {len(test_sentences)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def check_quota(activity_name, conn):
    cursor = conn.cursor()
    # 取得總名額
    cursor.execute("SELECT quota FROM activities WHERE name = ?", (activity_name,))
    row = cursor.fetchone()
    if row is None: return None, None
    total_quota = row[0]

    # 取得已報名人數
    cursor.execute("SELECT SUM(pax) FROM registrations WHERE activity_name = ? AND status = '已報名'", (activity_name,))
    sum_row = cursor.fetchone()
    used_quota = sum_row[0] if sum_row[0] else 0

    return total_quota, (total_quota - used_quota)

def register(name, activity, pax):
    if not name: return "請提供姓名。"
    if pax < 1 or pax > 3: return "報名人數必須在 1 到 3 人之間。"

    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()

        # 1. 檢查每人報名活動上限 (A2)
        cursor.execute("SELECT COUNT(DISTINCT activity_name) FROM registrations WHERE student_name = ? AND status = '已報名'", (name,))
        active_events = cursor.fetchone()[0]
        # 檢查是否已報名過此活動
        cursor.execute("SELECT 1 FROM registrations WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (name, activity))
        already_registered_this = cursor.fetchone() is not None

        if not already_registered_this and active_events >= 2:
            return "您已達每人最多同時報名 2 個活動的上限。"

        if already_registered_this:
            return "您已經報名過這個活動了，不可重複報名。"

        # 2. 檢查名額 (A1)
        total, remain = check_quota(activity, conn)
        if remain is None: return "找不到該活動。"
        if remain < pax: return f"名額不足！剩餘名額為 {remain} 人。"

        # 3. 寫入報名資料
        cursor.execute("INSERT INTO registrations (student_name, activity_name, pax, status) VALUES (?, ?, ?, '已報名')", (name, activity, pax))
        conn.commit()
        return f"報名成功！已為 {name} 報名 {activity} 共 {pax} 人。"
    finally:
        conn.close()

def cancel(name, activity):
    if not name: return "請提供姓名。"
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT id FROM registrations WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (name, activity))
        row = cursor.fetchone()
        if row is None:
            return "找不到您的報名紀錄。"

        cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (row[0],))
        conn.commit()
        return f"取消成功！已釋出 {activity} 的名額。"
    finally:
        conn.close()

def query(activity):
    conn = sqlite3.connect('activity_system.db')
    try:
        total, remain = check_quota(activity, conn)
        if remain is None: return "找不到該活動。"
        return f"【{activity}】剩餘名額為 {remain} 人。"
    finally:
        conn.close()

def stats():
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()
        # A3 統計查詢：JOIN 與 GROUP BY
        sql = '''
            SELECT a.date, a.name, a.quota, COALESCE(SUM(r.pax), 0) as used
            FROM activities a
            LEFT JOIN registrations r ON a.name = r.activity_name AND r.status = '已報名'
            GROUP BY a.id, a.name, a.date, a.quota
        '''
        cursor.execute(sql)
        rows = cursor.fetchall()

        result = "📊 各活動報名狀況：\n"
        for date, name, quota, used in rows:
            remain = quota - used
            result += f"- {date} {name}: 總額 {quota} / 已報 {used} / 剩餘 {remain}\n"
        return result
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
def process_request(name, text):
    action, activity, pax, raw = parse_intent_with_ai(text)

    if action == "未知":
        return "無法判斷您的需求，或者活動不存在。目前活動有：迎新晚會、程式工作坊、籃球友誼賽。\n您可以說「各活動報名狀況」查詢。"

    if action == "統計":
        return stats()

    if activity == "未知":
        return "無法判斷是哪一個活動。目前活動有：迎新晚會、程式工作坊、籃球友誼賽。"

    if action == "報名":
        return register(name, activity, pax)
    elif action == "取消":
        return cancel(name, activity)
    elif action == "查詢":
        return query(activity)

# 流程測試（依序執行）
print("==== 開始流程測試 ====")
init_db() # 測試前重置資料庫

steps = [
    ("王大明", "我和兩個朋友要報名迎新晚會"),
    ("王大明", "我要報名程式工作坊"),
    ("王大明", "我要報名籃球友誼賽"),
    ("李小華", "我要報名迎新晚會"),
    ("陳阿喵", "我和兩個朋友要報名迎新晚會"),
    ("王大明", "幫我取消程式工作坊"),
    ("李小華", "幫我取消籃球友誼賽"),
    ("林同學", "各活動報名狀況")
]

for i, (student, text) in enumerate(steps, 1):
    res = process_request(student, text)
    print(f"Step {i}: [{student}] 說：「{text}」")
    print(f"系統回應: {res}\n")

print("==== 最終資料表狀態 ====")
show_tables()
````
